```python
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("a.", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("b.", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("c.", name_list)

# d. Pop the last element from age_list
age_list.pop()
print("d.", age_list)

# e. Extend age_list with [29, 30, 26]
age_list.extend([29, 30, 26])
print("e.", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("f.", age_list)

# g. Find Max, Min and Sum
print("Max age:", max(age_list))
print("Min age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# a. Create a dictionary
student_marks = {
    "Anu": 75,
    "Rahul": 68,
    "Neha": 90,
    "Arjun": 85,
    "Priya": 78
}

print("a.", student_marks)

# b. Access and print the mark of a specific student
print("b. Mark of Neha:", student_marks["Neha"])

# c. Add a new student Janani with mark 80
student_marks["Janani"] = 80
print("c.", student_marks)

# d. Update the mark of one older student to 82
student_marks["Rahul"] = 82
print("d.", student_marks)

# e. Print keys, values, and items
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

```python
# a. Create a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("a.", my_set)

# Explanation
print("Explanation: Sets store only unique values, so duplicate elements are removed.")

# b. Attempt to change my_set[4] = 's'
# This will give an error because sets do not support indexing.

try:
    my_set[4] = 's'
except TypeError as e:
    print("b. Error:", e)
    print("Explanation: Sets are unordered and do not support indexing.")

# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set1:", set1)
print("Set2:", set2)

# d. Union and Intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

```python
# Performance Category Program

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a value between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent work! Keep it up.")

elif score >= 4 and score <= 7:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
